## Regression Linear
#### It predicts a continuous numerical value
## Linear regression: the simplest regression algorithm (y = mx+c)
#### It tries to predict a numerical value by fitting the best straight line through the data

```python
import pandas as pd # read and work on data
import numpy as np # numerical operations
import matplotlib.pyplot as plt # draw graphs           
```

### Read the file and print the first few rows of the csv file

```python
data = pd.read_csv("student_score.csv")# reads that file and converts it into dataframe(a tabel in python)
# This CSVsv file contains no. of hours each student studied and the marks scored 
data.head()
```

### Explore the Data

```python
data.info() #general info
```

```python
data.describe() # statistical summary
```

```python
plt.scatter(
    data["Hours_Studied"], #x-axis
    data["Marks"]) # y-axis
plt.xlabel("Hours_Studied")
plt.ylabel("Marks")
plt.title("Hours studied vs Marks")

plt.show()
```

```python
x = data[["Hours_Studied"]].values
y = data[["Marks"]].values
```

### create the model

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

### Train the model

```python
model.fit(x,y)
```

```python
plt.scatter(x,y,color="blue", label = "Actual Data")
plt.plot(x,model.predict(x),color="red", label = "regression Line")

plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Linear Regression")
plt.legend()

plt.show()
```

```python
print(model.coef_) #slope
```

```python
print(model.intercept_) #intercept
```

```python
prediction = model.predict([[5.5]])
print(prediction)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y, model.predict(x)) #measures how good the model is, r^2 = 1 means perfect and 0 means bad
print("r^2 score: " , r2)
```
